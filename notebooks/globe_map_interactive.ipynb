Define the function

```python
import plotly.graph_objects as go

def create_interactive_globe():
    """
    Creates an interactive globe using Plotly's Geo projection.
    """
    fig = go.Figure()

    # Add a scatter_geo trace (you can customize with your data points)
    fig.add_trace(
        go.Scattergeo(
            lon=[-74.006, 139.6917, 2.3522],  # Example longitudes (New York, Tokyo, Paris)
            lat=[40.7128, 35.6895, 48.8566],  # Example latitudes
            mode="markers",
            marker=dict(size=10, color="red"),
            text=["New York", "Tokyo", "Paris"],  # Labels for the points
        )
    )

    # Update layout for globe projection
    fig.update_layout(
        geo=dict(
            projection=dict(type="orthographic"),  # Globe-like projection
            showland=True,
            landcolor="rgb(243, 243, 243)",
            showocean=True,
            oceancolor="rgb(135, 206, 250)",
            showcountries=True,
            countrycolor="rgb(204, 204, 204)",
        ),
        title="Interactive Globe",
    )

    fig.show()
```

Example usage

```python
# Example usage
create_interactive_globe()
```
